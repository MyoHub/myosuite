### The `MyoSuite` imports needed for this tutorial

```python
import numpy as np

from myosuite import make_env  # importing myosuite registers the env ids
from myosuite.utils.video_io import show_video, write_video
```

### Load one of the `MyoSuite` tasks (see the complete set and descriptions [here](https://github.com/MyoHub/myosuite/blob/main/docs/source/environments.rst))

```python
env = make_env('myoElbowPose1D6MRandom-v0', render_mode='rgb_array')
```

### Generate some random movements
1. Reset the environment
2. Render the visualization of the environment
3. Send a random set of movements
4. Close the environment

```python
env.reset()
frames = []
for _ in range(100):
    frames.append(env.render())
    env.step(env.action_space.sample())  # take a random action
fps = round(1 / env.unwrapped.dt)  # one frame per control step: play back in real time
env.close()
```

### Save the frames from the rendering and play them in the notebook

```python
os.makedirs('videos', exist_ok=True)
# make a local copy
write_video('videos/temp.mp4', np.asarray(frames), fps=fps, outputdict={"-pix_fmt": "yuv420p"})

# show in the notebook
show_video('videos/temp.mp4')
```
