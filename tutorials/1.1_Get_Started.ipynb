# 1.1 — Load and Visualize a MyoSuite Model

Creates `myoElbowPose1D6MRandom-v0`, takes random actions, and renders an offscreen video.


```python
import os

#  If running on Colab, we'll install myosuite. If running locally, install these modules beforehand.
if os.getenv("COLAB_RELEASE_TAG"):
    !pip install git+https://github.com/MyoHub/myosuite

import gymnasium as gym
import myosuite  # registers env ids with Gymnasium
import numpy as np
from myosuite.utils.video_io import show_video, write_video

```

```python
env = gym.make('myoElbowPose1D6MRandom-v0', render_mode='rgb_array')
print('Cameras in this model:', [env.unwrapped.model.camera(i).name for i in range(env.unwrapped.model.ncam)])
env.reset()
frames = []
for _ in range(100):
    frames.append(env.render())
    env.step(env.action_space.sample())
env.close()

os.makedirs('videos', exist_ok=True)
# make a local copy
write_video('videos/temp.mp4', np.asarray(frames), outputdict={"-pix_fmt": "yuv420p"})

# show in the notebook
show_video('videos/temp.mp4')
```

```python
import matplotlib.pyplot as plt
plt.imshow(frames[0])
plt.axis('off')
plt.show()
```
